```python
import math
import pandas as pd
```

```python
df = pd.read_csv("data.csv")
df.head()
```

```python
k = 4
new_student = [0.50, 0.40]
distances = []
```

```python
for i, row in df.iterrows():
    distance = math.sqrt(
        (row["marks"] - new_student[0]) **2 + (row["attendance"] - new_student[1]) **2
    )
    distances.append([distance,row["result"]])
```

```python
distances.sort()
neighbors = distances[:k]
```

```python
count_0 = 0
count_1 = 0

for neighbor in neighbors:
    if neighbor[1] == 0:
        count_0 += 1
    else:
        count_1 += 1
```

```python
if count_1 > count_0:
    prediction = 1
else:
    prediction = 0

print("Prediction:", prediction)
```
